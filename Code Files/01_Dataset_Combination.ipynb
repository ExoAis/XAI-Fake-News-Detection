```python
import numpy as np
import pandas as pd
import json
```

## FakeNewsCorpus

```python
FNC = pd.read_csv('../Datasets/FakeNewsCorpus/news_sample.csv')
FNC = FNC[['type', 'content']]
```

## FakeNewsNet

```python
GSC_F = pd.read_csv('../Datasets/FakeNewsNet/dataset/gossipcop_fake.csv')
GSC_F.insert(1, 'type', ['fake' for i in range(GSC_F.shape[0])])
GSC_F = GSC_F[['title', 'type']]
GSC_R = pd.read_csv('../Datasets/FakeNewsNet/dataset/gossipcop_real.csv')
GSC_R.insert(1, 'type', ['real' for i in range(GSC_R.shape[0])])
GSC_R = GSC_R[['title', 'type']]
PTF_F = pd.read_csv('../Datasets/FakeNewsNet/dataset/politifact_fake.csv')
PTF_F.insert(1, 'type', ['fake' for i in range(PTF_F.shape[0])])
PTF_F = PTF_F[['title', 'type']]
PTF_R = pd.read_csv('../Datasets/FakeNewsNet/dataset/politifact_real.csv')
PTF_R.insert(1, 'type', ['real' for i in range(PTF_R.shape[0])])
PTF_R = PTF_R[['title', 'type']]
PTF_R.head()
FNN = pd.concat([GSC_F, GSC_R, PTF_F, PTF_R], ignore_index = True)
FNN = FNN.rename(columns = {'title' : 'content'})
FNN = FNN.sample(frac = 1).reset_index(drop = True)
```

## ISOT

```python
fake = pd.read_csv('../Datasets/ISOT/Fake.csv')
fake.insert(1, 'type', ['fake' for i in range(fake.shape[0])])
fake = fake[['text', 'type']]
real = pd.read_csv('../Datasets/ISOT/True.csv')
real.insert(1, 'type', ['real' for i in range(real.shape[0])])
real = real[['text', 'type']]
ISOT = pd.concat([fake, real], ignore_index = True)
ISOT = ISOT.rename(columns = {'text' : 'content'})
ISOT = ISOT.sample(frac = 1).reset_index(drop = True)
```

## LIAR-PLUS

```python
LP = pd.read_csv('../Datasets/LIAR-PLUS/dataset/tsv/train2.tsv', sep = '\t')
LP = LP.iloc[:, [15, 2]]
LP.rename(columns = {LP.columns[0] : 'content', LP.columns[1] : 'type'}, inplace = True)
```

## Politifact

```python
PF = []
with open('../Datasets/Politifact/politifact_factcheck_data.json', 'r') as file:
    for line in file:
        PF.append(json.loads(line))
PF_V = []
PF_S = []
for i in range(len(PF)):
    PF_V.append(PF[i]['verdict'])
    PF_S.append(PF[i]['statement'])
PF_V = np.array(PF_V)[:, np.newaxis]
PF_S = np.array(PF_S)[:, np.newaxis]
PF = np.concatenate([PF_V, PF_S], axis = 1)
PF = pd.DataFrame(data = PF, columns = ['type', 'content'])
```

## Combining

```python
Final = pd.concat([FNC, FNN, ISOT, LP, PF], ignore_index = True)
Final = Final.sample(frac = 1).reset_index(drop = True)
Final.to_csv('Datasets/Combined/Combined_Dataset.csv', index = False)
```
